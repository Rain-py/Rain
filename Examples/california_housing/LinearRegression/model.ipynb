```python
def get_train_data():
    return np.load("../../../data/california_housing/train_data.npy"), np.load("../../../data/california_housing/train_labels.npy")

def get_test_data():
    return np.load("../../../data/california_housing/test_data.npy"), np.load("../../../data/california_housing/test_labels.npy")
```

```python
X_train, y_train = get_train_data()
X_test, y_test = get_test_data()

scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
config = {
  "mode": {
      "type": "local",
      "params": {
        "num_of_workers": 3,
      }
  },
  "temp_data_path": "../../../Rain_data",
  "partitions": 3,
  "iterations": 50,
  "chunk_size": 1024*1024,
  "learning_type": "ML",
  "ML": {
      "algorithm": {
      "type": "LogisticRegression",
      "params": {
        "lr": 0.1
      }
    }    
  }
}
```

```python
# create linear regression model
model = LinearRegression(learning_rate=config["ML"]["algorithm"]["params"]["lr"], max_iters=config["iterations"])
model.fit(X_train, y_train)
```

```python
from sklearn import metrics
y_pred = model.predict(X_test)
print(f'R2 score: {metrics.r2_score(y_test, y_pred)}')
print(f'MSE: {metrics.mean_squared_error(y_test, y_pred)}')
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

print(f'R2 score: {metrics.r2_score(y_test, y_pred)}')
print(f'MSE: {metrics.mean_squared_error(y_test, y_pred)}')
```
