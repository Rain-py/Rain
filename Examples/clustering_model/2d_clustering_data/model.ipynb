```python
import numpy as np
import sys
sys.path.append('../../../src')
from Rain.Divider.MachineLearning.KMeans import KMeans as KMeans_rain
sys.path.pop()
```

```python
import os

def clean():
    folder_paths = ["logs"]  # Replace with the folder path you want to delete files from
    file_extensions = [".npy", ".pkl", ".log"]  # Replace with the file extension you want to delete
    for folder_path in folder_paths:
        if os.path.exists(folder_path):
            for filename in os.listdir(folder_path):
                for file_extension in file_extensions:
                    if filename.endswith(file_extension):
                        file_path = os.path.join(folder_path, filename)
                        os.remove(file_path)
clean()
```

```python
np.random.seed(63)
X = np.random.randn(250, 2)
```

```python
# create KMeans model
kmeans_model = KMeans_rain(n_clusters=3, max_iters=20)
labels = kmeans_model.fit(X)
```

```python
# Plot the data (X) along with the labels as colors
import matplotlib.pyplot as plt

for i in range(len(X)):
    if labels[i] == 0:
        plt.scatter(X[i, 0], X[i, 1], c='blue')
    elif labels[i] == 1:
        plt.scatter(X[i, 0], X[i, 1], c='green')
    elif labels[i] == 2:
        plt.scatter(X[i, 0], X[i, 1], c='red')

# Set labels for axes
plt.xlabel('x1')
plt.ylabel('x2')
plt.title('Rain Sequential KMeans Clustering')
plt.legend(['Cluster 1', 'Cluster 2', 'Cluster 3'])

# Show the plot
plt.show()
```

```python
# use sklearn kmeans
from sklearn.cluster import KMeans as KMeans_sklearn
kmeans_model_sklearn = KMeans_sklearn(n_clusters=3, max_iter=20)
labels_sklearn = kmeans_model_sklearn.fit_predict(X)
```

```python
for i in range(len(X)):
    if labels_sklearn[i] == 0:
        plt.scatter(X[i, 0], X[i, 1], c='red')
    elif labels_sklearn[i] == 1:
        plt.scatter(X[i, 0], X[i, 1], c='blue')
    elif labels_sklearn[i] == 2:
        plt.scatter(X[i, 0], X[i, 1], c='green')

# Set labels for axes
plt.xlabel('x1')
plt.ylabel('x2')
plt.legend(['Cluster 1', 'Cluster 2', 'Cluster 3'])
plt.title('Sklearn Sequential KMeans Clustering')

# Show the plot
plt.show()
```
