```python
import numpy as np
import sys
sys.path.append('../../../src')
from Rain.Divider.MachineLearning.KMeans import KMeans as KMeans_rain
sys.path.pop()
```

```python
import os

def clean():
    folder_paths = ["logs"]  # Replace with the folder path you want to delete files from
    file_extensions = [".npy", ".pkl", ".log"]  # Replace with the file extension you want to delete
    for folder_path in folder_paths:
        if os.path.exists(folder_path):
            for filename in os.listdir(folder_path):
                for file_extension in file_extensions:
                    if filename.endswith(file_extension):
                        file_path = os.path.join(folder_path, filename)
                        os.remove(file_path)
clean()
```

```python
np.random.seed(63)
X = np.random.randn(250, 2)
```

```python
# create KMeans model
kmeans_model = KMeans_rain(n_clusters=3, max_iters=20)
labels = kmeans_model.fit(X)
```

```python
# Plot the data (X) along with the labels as colors
import matplotlib.pyplot as plt

color_map = {0: 'red', 1: 'blue', 2: 'green'}

for label in set(labels):
    label_data = [d for d, l in zip(X, labels) if l == label]
    plt.scatter(*zip(*label_data), color=color_map[label], label=f"Cluster {label + 1}")


# Set labels for axes
plt.xlabel('x1')
plt.ylabel('x2')
plt.title('Rain Sequential K-means Clustering')

# Create the legend
plt.legend()

# Show the plot
plt.show()
```

```python
# use sklearn kmeans
from sklearn.cluster import KMeans as KMeans_sklearn
kmeans_model_sklearn = KMeans_sklearn(n_clusters=3, max_iter=20)
labels_sklearn = kmeans_model_sklearn.fit_predict(X)
```

```python
# Plot the data (X) along with the labels as colors
import matplotlib.pyplot as plt

color_map = {0: 'red', 1: 'blue', 2: 'green'}

for label in set(labels_sklearn):
    label_data = [d for d, l in zip(X, labels) if l == label]
    plt.scatter(*zip(*label_data), color=color_map[label], label=f"Cluster {label + 1}")


# Set labels for axes
plt.xlabel('x1')
plt.ylabel('x2')
plt.title('Sklearn K-means Clustering')

# Create the legend
plt.legend()
# Show the plot
plt.show()
```
