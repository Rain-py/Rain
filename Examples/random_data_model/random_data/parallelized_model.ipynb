```python
import os

def clean():
    folder_paths = ["logs"]  # Replace with the folder path you want to delete files from
    file_extensions = [".npy", ".pkl", ".log"]  # Replace with the file extension you want to delete
    for folder_path in folder_paths:
        if os.path.exists(folder_path):
            for filename in os.listdir(folder_path):
                for file_extension in file_extensions:
                    if filename.endswith(file_extension):
                        file_path = os.path.join(folder_path, filename)
                        os.remove(file_path)
clean()
```

```python
np.random.seed(63)
X = np.random.randn(250, 2)
```

```python
config = {
  "mode": {
      "type": "local",
      "params": {
        "num_of_workers": 3,
      }
  },
  "temp_data_path": "../../../",
  "partitions": 3,
  "iterations": 20,
  "chunk_size": 1024*1024,
  "learning_type": "ML",
  "ML": {
      "algorithm": {
      "type": "KMeans",
      "params": {
        "K": 3,
      }
    }    
  }
}
```

```python
rain = Rain(config)
```

```python
model = rain.train(X, None)
```

```python
labels = model.get_labels(X)
```

```python
# Plot the data (X) along with the labels as colors
import matplotlib.pyplot as plt

color_map = {0: 'red', 1: 'blue', 2: 'green'}

for label in set(labels):
    label_data = [d for d, l in zip(X, labels) if l == label]
    plt.scatter(*zip(*label_data), color=color_map[label], label=f"Cluster {label + 1}")


# Set labels for axes
plt.xlabel('x1')
plt.ylabel('x2')
plt.title('Rain Distributed K-means Clustering')

# Create the legend
plt.legend()
# Show the plot
plt.show()
```
