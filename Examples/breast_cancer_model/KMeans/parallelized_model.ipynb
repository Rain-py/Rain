```python
import os

def clean():
    folder_paths = ["logs"]  # Replace with the folder path you want to delete files from
    file_extensions = [".npy", ".pkl", ".log"]  # Replace with the file extension you want to delete
    for folder_path in folder_paths:
        if os.path.exists(folder_path):
            for filename in os.listdir(folder_path):
                for file_extension in file_extensions:
                    if filename.endswith(file_extension):
                        file_path = os.path.join(folder_path, filename)
                        os.remove(file_path)
clean()
```

```python
np.random.seed(63)
X = np.random.randn(200, 2)  # Assuming 100 data points with 2 features
# y = None
```

```python
config = {
  "mode": {
      "type": "local",
      "params": {}
  },
  "partitions": 3,
  "num_of_workers": 3,
  "iterations": 10,
  "learning_type": "ML",
  "ML": {
      "algorithm": {
      "type": "KMeans",
      "params": {
        "K": 2,
      }
    }    
  }
}
```

```python
# model = KMeans(config=config)
rain = Rain(config)
```

```python
model = rain.train(X, None)
```

```python
def calculate_distances(X, cluster_centers):
    n_clusters = cluster_centers.shape[0]
    n_samples = X.shape[0]
    distances = np.zeros((n_samples, n_clusters))

    for cluster in range(n_clusters):
        distances[:, cluster] = np.linalg.norm(X - cluster_centers[cluster], axis=1)

    return distances

distances = calculate_distances(X, model.cluster_centers)
labels = np.argmin(distances, axis=1)
```

```python
# Plot the data (X) along with the labels as colors
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

for i in range(len(X)):
    if labels[i] == 0:
        plt.scatter(X[i, 0], X[i, 1], c='red')
    elif labels[i] == 1:
        plt.scatter(X[i, 0], X[i, 1], c='blue')
    elif labels[i] == 2:
        plt.scatter(X[i, 0], X[i, 1], c='green')
    elif labels[i] == 3:
        plt.scatter(X[i, 0], X[i, 1], c='yellow')

# Set labels for axes
plt.xlabel('X')
plt.ylabel('Y')

# Show the plot
plt.show()
```
