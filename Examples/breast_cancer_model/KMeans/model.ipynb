```python
import os

def clean():
    folder_paths = ["logs"]  # Replace with the folder path you want to delete files from
    file_extensions = [".npy", ".pkl", ".log"]  # Replace with the file extension you want to delete
    for folder_path in folder_paths:
        if os.path.exists(folder_path):
            for filename in os.listdir(folder_path):
                for file_extension in file_extensions:
                    if filename.endswith(file_extension):
                        file_path = os.path.join(folder_path, filename)
                        os.remove(file_path)
clean()
```

```python
np.random.seed(63)
X = np.random.randn(200, 2)  # Assuming 100 data points with 2 features
```

```python
# create KMeans model
kmeans_model = KMeans_rain(n_clusters=2, max_iters=20)
labels = kmeans_model.fit(X)
```

```python
# Plot the data (X) along with the labels as colors
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

for i in range(len(X)):
    if labels[i] == 0:
        plt.scatter(X[i, 0], X[i, 1], c='red')
    elif labels[i] == 1:
        plt.scatter(X[i, 0], X[i, 1], c='blue')
    elif labels[i] == 2:
        plt.scatter(X[i, 0], X[i, 1], c='green')

# Set labels for axes
plt.xlabel('x1')
plt.ylabel('x2')

# Show the plot
plt.show()
```
