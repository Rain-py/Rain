```python
import numpy as np
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier as KNN_classifier_sklearn
# import naive_bayes_classifier from sklearn
from sklearn.naive_bayes import GaussianNB as naive_bayes_classifier_sklearn

```

```python
import os

def clean():
    folder_paths = ["logs"]  # Replace with the folder path you want to delete files from
    file_extensions = [".npy", ".pkl", ".log"]  # Replace with the file extension you want to delete
    for folder_path in folder_paths:
        if os.path.exists(folder_path):
            for filename in os.listdir(folder_path):
                for file_extension in file_extensions:
                    if filename.endswith(file_extension):
                        file_path = os.path.join(folder_path, filename)
                        os.remove(file_path)
clean()
```

```python
def get_train_data():
    return np.load("../../../data/breast_cancer/train_data.npy"), np.load("../../../data/breast_cancer/train_labels.npy")

def get_test_data():
    return np.load("../../../data/breast_cancer/test_data.npy"), np.load("../../../data/breast_cancer/test_labels.npy")
```

```python
X_train, y_train = get_train_data()
X_test, y_test = get_test_data()
```

```python
config = {
  "mode": "local",
  "partitions": 3,
  "num_of_workers": 3,
  "iterations": 3,
  "learning_type": "DL",
  "ML": {
      "algorithm": {
      "type": "KNN",
      "params": {
        "K": 5,
        "metric": "euclidean"
      }
    }    
  }
}
```

```python
# create KNN classifier
knn_classifier = KNN_classifier_sklearn(n_neighbors=config["ML"]["algorithm"]["params"]["K"], metric=config["ML"]["algorithm"]["params"]["metric"])

knn_classifier.fit(X_train, y_train)
y_pred_sklearn = knn_classifier.predict(X_test)
print(f"Accuracy of KNN classifier is: {(np.sum(y_pred_sklearn == y_test) / len(y_test) * 100).round(2)}%")
```

```python
# create naive bayes classifier
naive_bayes_classifier = naive_bayes_classifier_sklearn()
naive_bayes_classifier.fit(X_train, y_train)
y_pred_sklearn = naive_bayes_classifier.predict(X_test)
print(f"Accuracy of naive bayes classifier is: {(np.sum(y_pred_sklearn == y_test) / len(y_test) * 100).round(2)}%")
```
