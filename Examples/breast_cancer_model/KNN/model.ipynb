```python
def get_train_data():
    return np.load("../../../data/breast_cancer/train_data.npy"), np.load("../../../data/breast_cancer/train_labels.npy")

def get_test_data():
    return np.load("../../../data/breast_cancer/test_data.npy"), np.load("../../../data/breast_cancer/test_labels.npy")
```

```python
config = {
    "learning_type": "ML",
    "K": 5,
    "metric": "euclidean"
}
```

```python
X_train, y_train = get_train_data()
X_test, y_test = get_test_data()
```

```python
# create KNN classifier
knn_classifier = KNN_classifier_sklearn(n_neighbors=config["K"], metric=config["metric"])
knn_classifier.fit(X_train, y_train)
y_pred_sklearn = knn_classifier.predict(X_test)
print(f"Accuracy of KNN classifier is: {(np.sum(y_pred_sklearn == y_test) / len(y_test) * 100).round(2)}%")
```
